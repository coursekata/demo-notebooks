## 2.0: Model Variation with a Linear Function

**2.1a:** Find the best-fitting linear model.

```r
#run this code
lm(Consumer_Satisfaction ~ Pulse_Max, data = Haunted_House)
```

**2.1b:** Make a custom function from the best-fitting model and add it to the scatter plot.

```r
# modify this code 
linear_function <- function(X){0 + 0*X}

# add the function to the plot
gf_point(Consumer_Satisfaction ~ Pulse_Max, data = Haunted_House, color = "purple") %>%
  gf_function(linear_function)
```

**2.2:** How much error was explained by your model?

```r
# Modify and run this code: setting X and Y to our variables of interest
X <- Haunted_House$Pulse_Max
Y <- Haunted_House$Consumer_Satisfaction

# We will use these codes to calculate SSE(Sum of Squared Errors), 
# SST(Total - around the empty model), and PRE(Proportional Reduction in Error)

sse <- sse(Y, linear_function(X))
sst <- sse(Y, mean(Y))

(sst - sse) / sst
```

## 3.0: What's a polynomial function?

So far we have only focused on models where the most complex term has an $X$. But now, you get to add on $X^2$, $X^3$, $X^4$, etc.

```r
# edit this to see what happens with additional polynomial terms
poly_function <- function(X){ 0 + 1*X + -1*X^2 }

# create some simple X and Y variables
X <- c(-10:10)
Y <- poly_function(X)

# make a graph of this X and Y
gf_point(Y ~ X)
```

## 3 Explore Variation in `loveable`

### Introduction to Word Equations

We have listed our hypotheses about how an **explanatory variable** might help to predict a character's loveable levels (the **outcome variable**).

Now we will translate a hypothesis into a **word equation**, an informal model of the world. Word equations take the form: 

**Outcome Variable = Explanatory Variable + Other Stuff**

**Y = f(X) + Other Stuff**

### Example: Cool Characters!

For example, this hypothesis, *characters that are cool will be rated as more lovable* can be expressed like this: 

**loveable = cool + other stuff**

Or to use function notation, like this:

**loveable = f(cool) + other stuff**

## 4 Begin to Model Variation in `loveable`

"Modeling data" often means that we're 

We started with a word equation: 
- $Y = f(X) + \text{other stuff}$

But now let's formalize that into a linear equation
- Y = intercept + slope*X + other stuff
- or $Y = a + bX + \text{other stuff}$

**4.1, discussion:** Which part of the equation above is $f(X)$? 

**4.2:** We have added some code to add an "a-b-line", a line that has an intercept and a slope. Modify it until you feel like your line "fits" with this data best.

Consider: What were you using to judge whether this line "fit" this data well?

````r
gf_point(loveable ~ attractive, data = characters, color = "tomato") %>%
  gf_abline(intercept = 0, slope = .8)
````

**4.3:** Let's rewrite our word euqation to make a general purpose function. Modify the code below to make your line into a function.


````r
my_function <- function(X){ 0 + 0*X }

gf_point(loveable ~ attractive, data = characters, color = "tomato") %>%
  gf_function(my_function)
````

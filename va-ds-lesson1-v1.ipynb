## 12. Add a line like the one you drew on the paper to the scatter plot by modifying the intercept and slope in gf_abline.

```r
# modify the intercept and slope

gf_point(loveable ~ moody, data = characters, shape = 1) %>%
  gf_abline(intercept = 0, slope = 0)
```

## 14.  Maybe it will help to add the origin. Run this code to add a red dot to show the origin point.

```r
#run this code

gf_point(loveable ~ moody , data = characters, shape = 1) %>% 

  gf_abline(intercept = 95, slope = -0.8) %>%
 
  gf_point(0~0, color = "red")
```

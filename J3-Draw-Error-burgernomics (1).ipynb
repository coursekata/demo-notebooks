**4.1a:** What is a residual? How can we calculate them?

```r
#modify this code to find your residuals (replace "aleks_function" with the name of your function)
mutate(bigmac1, 
      prediction = aleks_function(year),
      residual = price - prediction)

#check your dataframe to see if your new variable is added
head(bigmac1)
```

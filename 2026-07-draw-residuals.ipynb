# The Tip Experiment

The `TipExperiment` is a data set from this [1996 paper](https://onlinelibrary.wiley.com/doi/abs/10.1111/j.1559-1816.1996.tb01847.x). In this experiment, 44 restaurant tables were randomly assigned to either receive smiley faces on their checks or not. The researchers hypothesized that tables would tip a higher percentage of the check if they got a hand-drawn smiley face on the check than if they did not.

```r
# Run this to see a sample of the data
sample(TipExperiment, 10)
```

## Modeling Variation in Tip 

Here we fit a model that uses `Condition` to model variation in `Tip` and depict it on a jitter plot.

```r
# Run this
Condition_model <- lm(Tip ~ Condition, data = TipExperiment)

gf_jitter(Tip ~ Condition, data = TipExperiment, width = .1) %>%
  gf_model(Condition_model)
```

## Depicting Residual Error from the Model Predictions

After students have experience gesturing residual error and drawing residual error, they can also use R code to depict residual error.

Thank you to Tyler Haslam, a CourseKata instructor and TIG-member for working on `gf_resid()` and `gf_square_resid()`.

```r
# Run this to draw on residual error 
gf_jitter(Tip ~ Condition, data = TipExperiment, width = .1) %>%
  gf_model(Condition_model) %>%
  gf_resid(Condition_model)
```

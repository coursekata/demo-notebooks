## Higher Degrees
What if we feel like the data is too cuvy for a linear model?  Maybe we think a higher degree polynomial is better!

### 5) Try fitting a degree two polynomial
**TIP: use `I(income^2)` to make a $x^2$ variable**

```r
# This code will make a model for Y = b_0 + b_1 X + b_2 X^2
deg2 <- lm(produce ~ income + I(income^2), data = Groceries)
deg2
```

```r
# can you make a third degree polynomial modal?  
deg3 <- lm(produce ~                  , data = Groceries)
deg3
```

## An easier way
There is a short cut for making polynomial models: `poly()`

### 6) run the code below to see how it works

```r
# Run this code to see poly() in action
deg3 <- lm(produce ~ poly(income, 3, raw = TRUE), data = Groceries)
deg3
```

```r
# To add the model to the graph we use gf_lm(), like this...
gf_jitter(produce ~ income, data = Groceries) %>%
  gf_lm(formula = y ~ poly(x, 3))
```

## The Best Fitting Model
Now that you know how to fit polynomial models in R, try finding the best fitting model

### 7) Find the polynomial model that you think best fits the data by modifying the code below

```r
# this sets the degree
DEG <- 3

# this code takes the degree you set and builds the model (no need to change it)
model <- lm(produce ~ poly(income, DEG ,raw=TRUE), data = Groceries)
model

# this takes your model and adds it to your graph
gf_jitter(produce ~ income, data = Groceries) %>%
  gf_lm(formula = y ~ poly(x, DEG ))

# and this gives you your model's error
supernova(model)
```
